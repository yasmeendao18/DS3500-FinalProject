```python
import plotly.express as px
import pandas as pd

# create sample data
df = pd.DataFrame({'x': [1, 2, 3, 4, 5], 'y1': [2, 3, 5, 4, 6], 'y2': [1, 2, 1, 3, 2], 'y3': [5, 4, 6, 7, 5]})

data = pd.read_csv("household_data.csv")
line_df = data.filter(regex='.*hyg.*|iso3|year|country', axis=1)
services = {
    'baseline': ['bas', 'lim', 'umimp', 'sur', 'nfac', 'od'],
    'available': ['sm', 'premises', 'available', 'quality', 'sdo_sm', 'fst_sm', 'sew_sm'],
    'infrastructure': ['pip', 'npip', 'lat', 'sep', 'sew'],
}

# remap service level names to more readable names
service_map = {
    'bas': 'Basic',
    'lim': 'Limited',
    'umimp': 'Unimproved',
    'sur': 'Surface water',
    'nfac': 'No facility',
    'od': 'Open defecation',
    'sm': 'Safe management',
    'premises': 'On premises',
    'available': 'Available when needed',
    'quality': 'Free from contamination',
    'sdo_sm': 'Disposed of',
    'fst_sm': 'Emptied and treated',
    'sew_sm': 'Waste water treated',
    'pip': 'Piped',
    'npip': 'Non-piped',
    'lat': 'Latrine',
    'sep': 'Septic tank',
    'sew': 'Sewerage system',
}
serv=services['baseline']
res_type = '_n'
fil_cols = ['iso3', 'year', 'country'] + [col for col in line_df.columns if col.endswith(res_type)]
line_df = line_df[fil_cols]
filtered_df = line_df[line_df["iso3"] == 'USA']
new_dict = {string: service_map[key] for string in filtered_df.columns for key in service_map.keys() if key in string}
print(fil_cols)
filtered_df = filtered_df.rename(columns=new_dict)
print(filtered_df.columns)


# build the line chart
line_fig = px.line(
    filtered_df,
    x="year",
    y=filtered_df.columns[3:],
    height=400,
)
line_fig
```
